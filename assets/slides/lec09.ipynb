# Mutability

```python
alist = [1, 2, 3, 4]
alist == [1, 2, 3, 4]
```

```python
print(alist == [1, 2, 3, 4, 5])
print(alist == [1, 2, 4, 3])
print(alist is [1, 2, 3, 4])
```

```python
blist = alist
alist is blist

alist.append(5)

print("alist: " , alist)
print("blist: " , blist)

print(alist == blist)
print(alist is blist)
```

```python
clist = list(blist)
clist

print(clist is blist)
```

```python
clist.append(6)
print(clist == blist)
print("blist: " , blist)
print("clist: ", clist)
```

```python
dlist = clist[:]
print(dlist is clist)
print(dlist == clist)
```

## Passing Data Into Functions

```python
def f(lst):
    lst = [99, 99]  # just rebinds the local name "lst"

my_list = [1, 2, 3]
f(my_list)
print(my_list)
```

```python
def f(lst):
    lst.append(99)  # directly modifies the object

my_list = [1, 2, 3]
f(my_list)
print(my_list)
```

```python
x = [1,2]
print(f"Id of x: {id(x)}")

x += [3]
print(f"Id of x: {id(x)}")

print("\n")

y = [1,2]
print(f"Id of x: {id(y)}")

y = y + [3]
print(f"Id of y: {id(y)}")
```
